# **Data Science: Exploración y análisis de datos**

## **Clase 1: Introducción**

```python
import pandas as pd
```

```python
notas = pd.read_csv("https://raw.githubusercontent.com/ahcamachod/2299-data-science-exploracion-y-analisis-de-datos/refs/heads/main/ratings.csv")
notas
```

```python
notas.shape
```

```python
notas.columns = ["usuarioId", "peliculaId", "nota", "momento"]
notas.head()
```

```python
notas["nota"].unique()
```

```python
notas["nota"].value_counts()
```

```python
notas["nota"].mean()
```

```python
notas["nota"].plot(kind='hist')
```

```python
notas["nota"].median()
```

```python
mediana = notas["nota"].median()
media = notas["nota"].mean()
print(f"Mediana é {mediana}")
print(f"Média é {media}")
```

```python
notas["nota"].describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(notas["nota"])
plt.show()
```

## Clase 2: Explorando las películas

```python
peliculas = pd.read_csv("https://raw.githubusercontent.com/ahcamachod/2299-data-science-exploracion-y-analisis-de-datos/refs/heads/main/movies.csv")
peliculas.columns = ["peliculaId", "titulo", "generos"]
peliculas.head()
```

```python
notas.head()
```

```python
notas.query("peliculaId==1")["nota"].mean()
```

```python
notas.query("peliculaId==2")["nota"].mean()
```

```python
promedio_por_pelicula = notas.groupby("peliculaId")["nota"].mean()
promedio_por_pelicula.head()
```

```python
promedio_por_pelicula.plot(kind="hist")
plt.show()
```

```python
sns.boxplot(promedio_por_pelicula)
plt.show()
```

```python
promedio_por_pelicula.describe()
```

```python
sns.displot(promedio_por_pelicula, kde=True)
plt.title("Histograma del promedio de notas de las películas")
plt.show()
```
