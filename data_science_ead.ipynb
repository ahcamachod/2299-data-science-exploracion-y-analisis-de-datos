## Clase 2: Explorando las películas

```python
peliculas = pd.read_csv("https://raw.githubusercontent.com/ahcamachod/2299-data-science-exploracion-y-analisis-de-datos/refs/heads/main/movies.csv")
peliculas.columns = ["peliculaId", "titulo", "generos"]
peliculas.head()
```

```python
notas.head()
```

```python
notas.query("peliculaId==1")["nota"].mean()
```

```python
notas.query("peliculaId==2")["nota"].mean()
```

```python
promedio_por_pelicula = notas.groupby("peliculaId")["nota"].mean()
promedio_por_pelicula.head()
```

```python
promedio_por_pelicula.plot(kind="hist")
plt.show()
```

```python
sns.boxplot(promedio_por_pelicula)
plt.show()
```

```python
promedio_por_pelicula.describe()
```

```python
sns.displot(promedio_por_pelicula, kde=True)
plt.title("Histograma del promedio de notas de las películas")
plt.show()
```

## Clase 3: Otras Películas

```python
tmdb = pd.read_csv("https://raw.githubusercontent.com/ahcamachod/2299-data-science-exploracion-y-analisis-de-datos/refs/heads/main/tmdb.csv")
tmdb.head()
```

```python
sns.displot(tmdb["revenue"])
plt.title("Distribución de ingresos de las películas")
plt.show()
```

```python
# prompt: gráfico de distribución del presupuesto para las películas (budget)

import matplotlib.pyplot as plt
sns.displot(tmdb["budget"])
plt.title("Distribución del presupuesto para las películas")
plt.show()
```

```python
tmdb.info()
```

```python
tmdb.describe()
```

```python
con_facturacion = tmdb.query("revenue > 0")
sns.displot(con_facturacion["revenue"])
plt.show()
```

```python
tmdb["original_language"].unique()
```

```python
tmdb["original_language"].value_counts()
```

```python
# idioma => categórica... sin orden...

# budget (presupuesto) => cuantitativo/continuo
```

```python
# nota de movielens => 0.5, 1, 1.5, ... ,5 => no tiene 2.5
# cantidad de votos => 1,2,3,4,5... no existe 2.5
```

```python
tmdb["original_language"].value_counts().index
```

```python
tmdb["original_language"].value_counts().values
```

```python
conteo_de_idioma = tmdb["original_language"].value_counts().to_frame().reset_index()
conteo_de_idioma.columns = ["original_language", "total"]
conteo_de_idioma.head()
```

```python
sns.barplot(data = conteo_de_idioma, x="original_language", y="total")
plt.show()
```

```python
sns.countplot(data=tmdb, x="original_language")
plt.show()
```

```python
# El gráfico de pizza -- Horroroso

conteo_de_idioma.plot(kind="pie", y="total", labels=conteo_de_idioma["original_language"])
plt.show()
```
