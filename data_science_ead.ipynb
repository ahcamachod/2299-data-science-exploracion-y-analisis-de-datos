## Clase 4: Mejorando nuestra visualización

```python
total_por_idioma = tmdb["original_language"].value_counts()
total_general = total_por_idioma.sum()
total_de_ingles = total_por_idioma.loc["en"]
total_de_otras = total_general - total_de_ingles
print(total_general, total_de_ingles, total_de_otras)
```

```python
datos = {
  "idioma" : ["Inglés", "Otros"],
  "total" : [total_de_ingles, total_de_otras]
}
datos = pd.DataFrame(datos)
datos
```

```python
sns.barplot(data=datos, x="idioma", y="total")
plt.show()
```

```python
datos.plot(kind="pie", y="total", labels=datos["idioma"])
plt.show()
```

```python
total_de_otras_peliculas_por_idioma = tmdb.query("original_language != 'en'")["original_language"].value_counts()
total_de_otras_peliculas_por_idioma.head()
```

```python
sns.countplot(data=tmdb.query("original_language != 'en'"),
              x="original_language")
plt.show()
```

```python
sns.countplot(data=tmdb.query("original_language != 'en'"),
              order=total_de_otras_peliculas_por_idioma.index,
              hue="original_language",
              x="original_language")
plt.show()
```

```python
sns.color_palette("mako")
```

```python
tmdb.query("original_language != 'en'")["original_language"].value_counts(normalize=True)
```

```python
plt.figure(figsize=(16, 8))
sns.countplot(data=tmdb.query("original_language != 'en'"),
              order=total_de_otras_peliculas_por_idioma.index,
              palette="mako",
              hue="original_language",
              hue_order=total_de_otras_peliculas_por_idioma.index,
              stat="percent",
              x="original_language")
plt.title("Distribución del idioma original en las peliculas excluyendo el inglés")
plt.show()
```

## Clase 5: Comparar dos distribuciones visualmente o a través de medidas

```python
peliculas.head(2)
```

```python
# prompt: Extrae las notas de dos películas en variables distintas
notas_de_toy_story = notas.query("peliculaId==1")["nota"]
notas_de_jumanji = notas.query("peliculaId==2")["nota"]

media_de_toy_story = notas_de_toy_story.mean()
media_de_jumanji = notas_de_jumanji.mean()

print(media_de_toy_story, media_de_jumanji)
```

```python
# prompt: lo mismo, pero con la mediana

mediana_de_toy_story = notas_de_toy_story.median()
mediana_de_jumanji = notas_de_jumanji.median()

print(mediana_de_toy_story, mediana_de_jumanji)
```

```python
import numpy as np

pelicula1 = [2.5] * 10 + [3.5] * 10
pelicula2 = [5] * 10 + [1] * 10

```

```python

media_pelicula1 = np.mean(pelicula1)
mediana_pelicula1 = np.median(pelicula1)

media_pelicula2 = np.mean(pelicula2)
mediana_pelicula2 = np.median(pelicula2)

print("pelicula 1:")
print("Média:", media_pelicula1)
print("Mediana:", mediana_pelicula1)

print("\npelicula 2:")
print("Média:", media_pelicula2)
print("Mediana:", mediana_pelicula2)
```

```python
plt.hist(pelicula1)
plt.hist(pelicula2)
plt.show()
```

```python
plt.boxplot([pelicula1, pelicula2])
plt.show()
```

```python
np.std(pelicula0), np.std(pelicula1), np.std(pelicula2)
```
